```python
import os
from dotenv import load_dotenv
import pandas as pd
```

```python
load_dotenv("../.env")
data_path = os.getenv("accepted_data_path")
```

```python
cols = ["id", "loan_amnt", "term", "int_rate", "grade", "emp_length",
    "home_ownership", "annual_inc", "verification_status", "purpose",
    "dti", "delinq_2yrs", "earliest_cr_line", "inq_last_6mths",
    "open_acc", "pub_rec", "revol_bal", "revol_util", "total_acc",
    "issue_d", "loan_status"]
```

```python
df = pd.read_csv(data_path, usecols=cols, low_memory=False)
```

```python
keep = ["Fully Paid", "Charged Off", "Default"]
df = df[df["loan_status"].isin(keep)].copy()
```

```python
df["target"] = (df["loan_status"] != "Fully Paid").astype(int)
```

```python
df["issue_d"] = pd.to_datetime(df["issue_d"], format="%b-%Y")
df["year"] = df["issue_d"].dt.year
```
